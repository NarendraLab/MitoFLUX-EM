```python
# Jupyter cell — same as previous sorted / no-wedge script but add median_color_hex per label to CSV.
import os, math, json, time, sys
import numpy as np
import pandas as pd
import tifffile
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib import cm
from matplotlib.colors import Normalize, ListedColormap, to_hex

# --------------- Config ---------------
INPUT_DIR = "./input"
OUT_DIR = "./results"
os.makedirs(OUT_DIR, exist_ok=True)

MAX_PLOT_DIM = 100          # downsample threshold for plotting crops
EXPORT_DPI = 300            # high-res output for Illustrator-friendly PDFs

# Default color scale requested
GLOBAL_VMIN = 0.0037
GLOBAL_VMAX = 0.6

# reproducible jitter for scatter in comparison plot
RNG = np.random.default_rng(42)

# Try to load custom FIJI LUT if available
def load_fiji_lut_try(paths=("LUT.csv", os.path.join(INPUT_DIR,"LUT.csv"))):
    for p in paths:
        try:
            if not os.path.exists(p):
                continue
            df = pd.read_csv(p)
            cols = [c.strip().lower() for c in df.columns]
            if "red" in cols and "green" in cols and "blue" in cols:
                r_idx = cols.index("red"); g_idx = cols.index("green"); b_idx = cols.index("blue")
            elif "r" in cols and "g" in cols and "b" in cols:
                r_idx = cols.index("r"); g_idx = cols.index("g"); b_idx = cols.index("b")
            else:
                r_idx, g_idx, b_idx = 1,2,3
            rgb = df.iloc[:, [r_idx, g_idx, b_idx]].to_numpy(dtype=float) / 255.0
            rgb = np.clip(rgb, 0.0, 1.0)
            cmap = ListedColormap(rgb, name="fiji_lut_from_csv")
            try:
                mpl.colormaps.register(cmap, name="fiji_lut_from_csv", force=True)
            except Exception:
                mpl.cm.register_cmap(name="fiji_lut_from_csv", cmap=cmap)
            print(f"Loaded LUT from {p} ({rgb.shape[0]} entries).")
            return cmap
        except Exception as e:
            print(f"Failed to load LUT from {p}: {e}", file=sys.stderr)
    return None

CUSTOM_CMAP = load_fiji_lut_try()
CMAP = CUSTOM_CMAP if CUSTOM_CMAP is not None else cm.get_cmap("viridis")
NORM_FOR_HEX = Normalize(vmin=GLOBAL_VMIN, vmax=GLOBAL_VMAX)

# --------------- Helpers ---------------
def read_stack(path):
    arr = tifffile.imread(path)
    if arr.ndim == 2:
        arr = arr[np.newaxis, ...]
    return arr

def auto_detect_files(input_dir):
    files = [f for f in os.listdir(input_dir) if f.lower().endswith((".tif", ".tiff"))]
    if len(files) == 0:
        raise FileNotFoundError(f"No .tif/.tiff files found in {input_dir}")
    if len(files) == 1:
        raise FileNotFoundError(f"Only one tiff found in {input_dir}. Expected two (ratio + labels). Found: {files}")
    if len(files) > 2:
        keywords = ["ratio","label","labels","content","mito","mims","mitolysosome"]
        ranked = sorted(files, key=lambda f: sum(k in f.lower() for k in keywords), reverse=True)
        files = ranked[:2]
    return [os.path.join(input_dir, f) for f in files[:2]]

def pick_ratio_and_label(pathA, pathB):
    a = tifffile.imread(pathA); b = tifffile.imread(pathB)
    if a.ndim == 2: a = a[np.newaxis,...]
    if b.ndim == 2: b = b[np.newaxis,...]
    if np.issubdtype(a.dtype, np.integer) and np.issubdtype(b.dtype, np.floating):
        return pathB, pathA
    if np.issubdtype(b.dtype, np.integer) and np.issubdtype(a.dtype, np.floating):
        return pathA, pathB
    def uniq_count(arr):
        flat = arr.ravel()
        if flat.size > 50000:
            sample = flat[np.random.choice(flat.size, 50000, replace=False)]
        else:
            sample = flat
        return np.unique(sample).size
    if uniq_count(a) < uniq_count(b):
        return pathB, pathA
    return pathA, pathB

def bbox_of_mask(mask):
    coords = np.where(mask)
    if coords[0].size == 0:
        return None
    z0,z1 = coords[0].min(), coords[0].max()+1
    y0,y1 = coords[1].min(), coords[1].max()+1
    x0,x1 = coords[2].min(), coords[2].max()+1
    return (z0,z1,y0,y1,x0,x1)

def downsample(arr, factor):
    if factor <= 1:
        return arr
    return arr[::factor, ::factor, ::factor]

def normalize_vals(vals, vmin, vmax):
    if math.isclose(vmin, vmax):
        return np.full_like(vals, 0.5, dtype=float)
    return (vals - vmin) / (vmax - vmin)

def add_colorbar_for_axes(ax, vmin, vmax, cmap=CMAP, label="fractional incorporation"):
    sm = cm.ScalarMappable(norm=Normalize(vmin=vmin, vmax=vmax), cmap=cmap)
    sm.set_array([])
    plt.colorbar(sm, ax=ax, fraction=0.046, pad=0.04, label=label)

def freedman_diaconis_bins(data):
    data = np.asarray(data)
    data = data[np.isfinite(data)]
    n = data.size
    if n == 0:
        return 10
    q75, q25 = np.percentile(data, [75, 25])
    iqr = q75 - q25
    if iqr == 0:
        return "auto"
    bin_width = 2 * iqr / (n ** (1/3))
    if bin_width <= 0:
        return "auto"
    bins = max(3, int(np.ceil((data.max() - data.min()) / bin_width)))
    return bins

# --------------- Auto-detect inputs & load ---------------
file_paths = auto_detect_files(INPUT_DIR)
print("Found TIFFs:", file_paths)
ratio_file, labels_file = pick_ratio_and_label(file_paths[0], file_paths[1])
print("Using ratio file:", ratio_file)
print("Using labels file:", labels_file)

ratio = read_stack(ratio_file).astype(float)   # (Z,Y,X)
labels = read_stack(labels_file).astype(int)
if ratio.shape != labels.shape:
    raise ValueError(f"Shape mismatch between ratio {ratio.shape} and labels {labels.shape}.")
print("Volume shape (Z, Y, X):", ratio.shape)

# --------------- Compute fractional incorporation and per-label summary stats ---------------
frac = ratio / (1.0 + ratio)   # fractional incorporation
labeled_mask = (labels != 0)
if labeled_mask.sum() == 0:
    raise ValueError("No labeled voxels found (labels appear all zero).")

unique_labels = np.unique(labels)
unique_labels = unique_labels[unique_labels != 0]
print("Labels found:", unique_labels.tolist())

records = []
# store per-label arrays for comparison plot (full-res)
perlabel_frac_vals = {}
start_time = time.time()

for lab in unique_labels:
    mask = (labels == lab)
    vals_ratio = ratio[mask]
    vals_frac = frac[mask]
    if vals_frac.size == 0:
        continue
    q25 = float(np.percentile(vals_frac, 25))
    q75 = float(np.percentile(vals_frac, 75))
    iqr = q75 - q25
    frac_median = float(np.median(vals_frac))

    # compute color hex for median fractional value (map median to colormap using GLOBAL range)
    norm_m = NORM_FOR_HEX(frac_median)
    norm_m = float(np.clip(norm_m, 0.0, 1.0))
    rgba = CMAP(norm_m)
    hexcol = to_hex(rgba)  # returns '#rrggbb' (alpha ignored)

    rec = {
        "label": int(lab),
        "count": int(mask.sum()),
        "ratio_median": float(np.median(vals_ratio)),
        "ratio_range": float(np.max(vals_ratio) - np.min(vals_ratio)),
        "ratio_std": float(np.std(vals_ratio)),
        "ratio_min": float(np.min(vals_ratio)),
        "ratio_max": float(np.max(vals_ratio)),
        "frac_median": frac_median,
        "frac_q25": q25,
        "frac_q75": q75,
        "frac_iqr": iqr,
        "frac_std": float(np.std(vals_frac)),
        "frac_min": float(np.min(vals_frac)),
        "frac_max": float(np.max(vals_frac)),
        "median_color_hex": hexcol,
        "values_ratio": json.dumps(vals_ratio.tolist()),
        "values_fractional": json.dumps(vals_frac.tolist())
    }
    records.append(rec)
    perlabel_frac_vals[int(lab)] = vals_frac

# build dataframe and sort labels by frac_median descending (highest first)
df = pd.DataFrame(records)
if df.empty:
    raise RuntimeError("No labeled data to process.")
df_sorted = df.sort_values(by="frac_median", ascending=False)
labels_sorted_by_median = df_sorted["label"].to_numpy(dtype=int)
print("Labels sorted by frac_median (highest -> lowest):", labels_sorted_by_median.tolist())

# save CSV (unsorted full table and also sorted)
csv_out = os.path.join(OUT_DIR, "label_stats_ratio_and_fractional_with_iqr_and_hex.csv")
df.to_csv(csv_out, index=False)
csv_sorted_out = os.path.join(OUT_DIR, "label_stats_ratio_and_fractional_with_iqr_and_hex_sorted.csv")
df_sorted.to_csv(csv_sorted_out, index=False)
print("Saved CSVs:", csv_out, csv_sorted_out)

# --------------- Plot labels in sorted order ---------------
for lab in labels_sorted_by_median:
    t0 = time.time()
    lab = int(lab)
    mask = (labels == lab)
    count = int(mask.sum())
    print(f"Plotting label {lab} (voxels={count})")

    # bbox + crop
    bb = bbox_of_mask(mask)
    if bb is None:
        print(f"Label {lab} empty bbox; skipping.")
        continue
    z0,z1,y0,y1,x0,x1 = bb
    mask_crop = mask[z0:z1, y0:y1, x0:x1]
    frac_crop = frac[z0:z1, y0:y1, x0:x1]

    # downsample for plotting if needed
    dz,dy,dx = mask_crop.shape
    max_dim = max(dz,dy,dx)
    ds_factor = 1
    if max_dim > MAX_PLOT_DIM:
        ds_factor = math.ceil(max_dim / MAX_PLOT_DIM)
        mask_ds = downsample(mask_crop, ds_factor)
        frac_ds = downsample(frac_crop, ds_factor)
    else:
        mask_ds = mask_crop
        frac_ds = frac_crop

    # Build voxel facecolors from fractional values using GLOBAL_VMIN/GLOBAL_VMAX
    mask_vox = mask_ds.transpose(2,1,0)   # (X,Y,Z)
    frac_vox = frac_ds.transpose(2,1,0)
    facecolors = np.zeros(mask_vox.shape + (4,), dtype=float)
    if mask_vox.any():
        vals_mask_frac = frac_vox[mask_vox]
        normed = normalize_vals(vals_mask_frac, GLOBAL_VMIN, GLOBAL_VMAX)
        normed = np.clip(normed, 0.0, 1.0)
        facecolors[mask_vox] = CMAP(normed)

    # --- Voxel plotting (full) ---
    fig = plt.figure(figsize=(6,6))
    ax = fig.add_subplot(111, projection='3d')
    ax.set_box_aspect(mask_vox.shape)
    ax.voxels(mask_vox, facecolors=facecolors, edgecolor=None)
    ax.set_title(f"Label {lab} — voxels (fractional color)\nvoxels={count}, ds={ds_factor}")
    ax.axis("off")
    add_colorbar_for_axes(ax, GLOBAL_VMIN, GLOBAL_VMAX, cmap=CMAP, label="fractional incorporation")
    out_full = os.path.join(OUT_DIR, f"label_{lab:03d}_voxels_fractional_sorted.pdf")
    fig.savefig(out_full, dpi=EXPORT_DPI, bbox_inches='tight', format='pdf')
    plt.show(); plt.close(fig)

    # --- MIPs and histogram (no wedge) ---
    inten_masked_ds_frac = np.where(mask_ds, frac_ds, np.nan)  # (Z,Y,X)
    top_mip = np.nanmax(inten_masked_ds_frac, axis=0)    # Y x X
    front_mip = np.nanmax(inten_masked_ds_frac, axis=1)  # Z x X
    side_mip = np.nanmax(inten_masked_ds_frac, axis=2)   # Z x Y

    # fill NaNs for display with a value slightly below vmin so missing rays appear distinct
    fill_val = GLOBAL_VMIN - 0.02 * (GLOBAL_VMAX - GLOBAL_VMIN)
    def fillnan(arr):
        a = np.array(arr, dtype=float)
        a[np.isnan(a)] = fill_val
        return a

    top_img = fillnan(top_mip)
    front_img = fillnan(front_mip)
    side_img = fillnan(side_mip)

    # histogram (full-res values used for histogram & CSV)
    vals_frac_full = perlabel_frac_vals[lab]
    bins = freedman_diaconis_bins(vals_frac_full)
    hist_bins = 'auto' if bins == "auto" else bins

    # combined panel: Top | Front | Side | Histogram (with reserved scalebar axis)
    fig = plt.figure(figsize=(20,6))
    gs = fig.add_gridspec(1, 6, width_ratios=[1,1,1,1,0.12,0.12], wspace=0.12)
    ax_top = fig.add_subplot(gs[0,0])
    ax_front = fig.add_subplot(gs[0,1])
    ax_side = fig.add_subplot(gs[0,2])
    ax_hist = fig.add_subplot(gs[0,3])
    ax_scale = fig.add_subplot(gs[0,4])
    ax_pad = fig.add_subplot(gs[0,5])

    ax_top.imshow(top_img, cmap=CMAP, vmin=GLOBAL_VMIN, vmax=GLOBAL_VMAX, origin='lower', aspect='equal')
    ax_top.set_title(f"Label {lab} — Top (Z→)"); ax_top.axis('off')

    ax_front.imshow(front_img, cmap=CMAP, vmin=GLOBAL_VMIN, vmax=GLOBAL_VMAX, origin='lower', aspect='equal')
    ax_front.set_title(f"Label {lab} — Front (Y→)"); ax_front.axis('off')

    ax_side.imshow(side_img, cmap=CMAP, vmin=GLOBAL_VMIN, vmax=GLOBAL_VMAX, origin='lower', aspect='equal')
    ax_side.set_title(f"Label {lab} — Side (X→)"); ax_side.axis('off')

    # histogram (full-res)
    x_range = (GLOBAL_VMIN, GLOBAL_VMAX)
    counts, edges, patches = ax_hist.hist(vals_frac_full, bins=hist_bins, range=x_range, density=False, alpha=0.95)
    bin_centers = 0.5 * (edges[:-1] + edges[1:])
    if bin_centers.size > 0:
        norm_centers = normalize_vals(bin_centers, GLOBAL_VMIN, GLOBAL_VMAX)
        norm_centers = np.clip(norm_centers, 0.0, 1.0)
        colors = CMAP(norm_centers)
        for p, c in zip(patches, colors):
            p.set_facecolor(c); p.set_edgecolor('none')

    med = float(np.median(vals_frac_full)); mu = float(np.mean(vals_frac_full))
    ax_hist.axvline(med, color='k', linestyle='--', label=f"median={med:.3g}")
    ax_hist.axvline(mu, color='r', linestyle=':', label=f"mean={mu:.3g}")
    ax_hist.set_title(f"Fractional Histogram — Label {lab} (n={vals_frac_full.size})")
    ax_hist.set_xlabel("fractional incorporation"); ax_hist.set_ylabel("count")
    ax_hist.legend(fontsize='small')

    # shared colorbar for MIPs
    sm = cm.ScalarMappable(cmap=CMAP, norm=Normalize(vmin=GLOBAL_VMIN, vmax=GLOBAL_VMAX))
    sm.set_array([])
    plt.colorbar(sm, ax=[ax_top, ax_front, ax_side], fraction=0.046, pad=0.02, label='fractional incorporation')

    # Scale-bar placed in reserved axis to avoid overlap
    crop_width_px = x1 - x0
    crop_height_px = y1 - y0
    sb_len_px = max(5, int(min(crop_width_px, crop_height_px) / 6.0))
    sb_label = f"{sb_len_px} px"
    ax_scale.axis('off')
    ax_scale.set_xlim(0,1); ax_scale.set_ylim(0,1)
    ax_scale.plot([0.1, 0.1 + 0.8 * (sb_len_px / max(crop_width_px,1))], [0.5, 0.5], color='k', linewidth=4, solid_capstyle='butt')
    ax_scale.text(0.5, 0.35, sb_label, ha='center', va='center', fontsize=10)
    ax_pad.axis('off')

    out_combined = os.path.join(OUT_DIR, f"label_{lab:03d}_mips_and_frac_hist_sorted.pdf")
    fig.savefig(out_combined, dpi=EXPORT_DPI, bbox_inches='tight', format='pdf')
    plt.show(); plt.close(fig)

    print(f"Saved label {lab} figures (ds={ds_factor}) in {time.time()-t0:.2f}s")

# --------------- Comparison plot ordered by median ---------------
labels_cmp = labels_sorted_by_median
frac_lists_sorted = [perlabel_frac_vals[int(l)] for l in labels_cmp]
x_positions = np.arange(len(labels_cmp))

fig, ax = plt.subplots(figsize=(max(8, len(labels_cmp)*0.5), 6))

# individual voxel points (jittered) colored by fractional value using global vmin/vmax
for xi, vals in zip(x_positions, frac_lists_sorted):
    if vals.size == 0:
        continue
    jitter = (RNG.random(vals.size) - 0.5) * 0.6
    xs = np.full(vals.size, xi) + jitter
    cols = CMAP(np.clip((vals - GLOBAL_VMIN) / (GLOBAL_VMAX - GLOBAL_VMIN), 0.0, 1.0))
    ax.scatter(xs, vals, s=6, color=cols, alpha=0.7, edgecolors='none')

# medians and IQRs
medians = np.array([np.median(v) for v in frac_lists_sorted])
q25s = np.array([np.percentile(v,25) for v in frac_lists_sorted])
q75s = np.array([np.percentile(v,75) for v in frac_lists_sorted])

for xi, ql, qu in zip(x_positions, q25s, q75s):
    ax.plot([xi, xi], [ql, qu], color='k', linewidth=2, solid_capstyle='butt')

ax.plot(x_positions, medians, marker='o', linestyle='None', color='white', markeredgecolor='k', markersize=6, label='median')

ax.set_xticks(x_positions)
ax.set_xticklabels([str(int(l)) for l in labels_cmp], rotation=90)
ax.set_xlim(-0.5, len(x_positions)-0.5)
ax.set_ylim(GLOBAL_VMIN - 0.02, GLOBAL_VMAX + 0.02)
ax.set_ylabel("15N fractional incorporation")
ax.set_xlabel("Label (sorted by median frac: highest → lowest)")
ax.set_title("Per-label fractional incorporation — individual voxels (dots), IQR (vertical), median (circle)")

# colorbar for point colors
sm = cm.ScalarMappable(cmap=CMAP, norm=Normalize(vmin=GLOBAL_VMIN, vmax=GLOBAL_VMAX))
sm.set_array([])
plt.colorbar(sm, ax=ax, fraction=0.04, pad=0.02, label='fractional incorporation')

plt.tight_layout()
out_cmp = os.path.join(OUT_DIR, "labels_frac_median_iqr_and_points_sorted.pdf")
fig.savefig(out_cmp, dpi=EXPORT_DPI, bbox_inches='tight', format='pdf')
plt.show(); plt.close(fig)

print("Saved comparison plot:", out_cmp)
print("All outputs saved in:", os.path.abspath(OUT_DIR))
print("Done. Total time: %.1f s" % (time.time() - start_time))
```
